# ⚛️ Governed Quantum Agents: Molecule Discovery Example (IBM Qiskit)

This notebook demonstrates the underlying process used by the agents to discover molecules (e.g., finding candidate molecules that inhibit a specific enzyme). It simulates the step-by-step creation and execution of a quantum circuit using **IBM Qiskit**, mirroring what happens under the hood when you use the CLI.

```python
import sys
import os
import logging

# Ensure we can import from the src directory
sys.path.append(os.path.abspath('../src'))

from bridge.circuit_generator import CircuitGenerator
from quantum.execution import QuantumExecution

# Setup basic logging to see internal agent messages
logging.basicConfig(level=logging.INFO, format='%(levelname)s - %(message)s')
```

## 1. Problem Encoding (Simulated Agent Output)
When a user asks: *"Find a candidate molecule that inhibits BACE-1"*, the **LLM Planner** and **Problem Encoder** translate this natural language request into a quantum-compatible structure.

Here is a mock of what the AI generates:

```python
encoded_problem = {
    "molecule": "BACE-1 inhibitor candidate",
    "qubits_required": 2,
    "ansatz_layers": 1
}

# The Algorithm Selector Agent decides to use VQE for molecular simulation
selected_algorithm = "VQE"
print(f"Selected Algorithm: {selected_algorithm}")
```

## 2. Quantum Circuit Generation (Qiskit)
Next, the **Classical-Quantum Bridge** translates the encoded problem into a parameterized quantum circuit using IBM Qiskit.

```python
circuit_generator = CircuitGenerator()

# Let's generate a Qiskit circuit for this problem
circuit = circuit_generator.generate_circuit(selected_algorithm, encoded_problem, framework="qiskit")

print("\nGenerated VQE Ansatz Circuit (Qiskit):")
print("-" * 40)
print(circuit)
print("-" * 40)
```

## 3. Quantum Execution (Aer Simulator)
The generated circuit is sent to the **Quantum Execution Layer**, where it runs on a local simulator (`AerSimulator`) or real quantum hardware via the cloud.

```python
quantum_executor = QuantumExecution()

print("Executing the quantum simulation...")
execution_summary = quantum_executor.execute_circuit(circuit, repetitions=1000)

print("\nExecution Results:")
print(f"Success: {execution_summary['success']}")
print(f"Repetitions: {execution_summary['repetitions']}")
print(f"Dominant State: {execution_summary['dominant_state']}")
print(f"Dominant Probability: {execution_summary['dominant_probability']:.2%}")
print("\nBitstring Counts:")
for bitstring, count in execution_summary['bitstring_counts'].items():
    print(f"  |{bitstring}\u27e9 : {count} times")
```

## 4. Interpretation
Finally, the **Interpreter Agent** reads these raw probability distributions (`bitstring_counts`) and translates them back into scientific English (e.g., mapping the dominant state to an estimated ground state energy and presenting the binding affinity score to the human researcher).
